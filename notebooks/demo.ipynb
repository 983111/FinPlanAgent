# FinPlanAgent Demo
## AI-Driven Personal Financial Advisor

This notebook demonstrates the complete workflow of FinPlanAgent.

```python
# Import required libraries
import sys
sys.path.append('..')

from src.parser import TransactionParser
from src.analyzer import TransactionAnalyzer
from src.profile_builder import ProfileBuilder
from src.optimizer import BudgetOptimizer
from src.agent import FinancialAgent
from src.simulator import WhatIfSimulator
from src.explainer import Explainer
from src.utils import plot_expense_breakdown, plot_trend

import os
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

print("✅ All modules imported successfully!")
```

## Step 1: Parse Transaction Data

```python
# Initialize parser
parser = TransactionParser()

# Parse sample transactions
df = parser.parse_csv('../data/sample_transactions.csv')

print(f"Loaded {len(df)} transactions")
print("\nFirst few transactions:")
df.head(10)
```

```python
# Validate data
validation = parser.validate_data(df)
print("Validation Results:")
print(f"Valid: {validation['valid']}")
print(f"\nStats:")
for key, value in validation['stats'].items():
    print(f"  {key}: {value}")
```

## Step 2: Analyze Transactions

```python
# Initialize analyzer
analyzer = TransactionAnalyzer()

# Categorize transactions
df_categorized = analyzer.categorize(df)

print("Transaction categories:")
print(df_categorized['category'].value_counts())
```

```python
# Compute spending patterns
patterns = analyzer.compute_spending_patterns(df_categorized)

print("\nSpending Patterns:")
print(f"Total Income: ₹{patterns['total_income']:,.2f}")
print(f"Total Expenses: ₹{patterns['total_expenses']:,.2f}")
print(f"Net Savings: ₹{patterns['net_savings']:,.2f}")
print(f"\nTop Spending Categories:")
for item in patterns['top_spending']:
    print(f"  {item['category']}: ₹{item['total']:,.2f}")
```

```python
# Visualize expense breakdown
fig = plot_expense_breakdown(patterns['by_category'])
plt.show()
```

```python
# Generate insights
insights = analyzer.generate_insights(df_categorized)

print("\n💡 Key Insights:")
for insight in insights:
    print(f"  • {insight}")
```

## Step 3: Build Financial Profile

```python
# Initialize profile builder
profile_builder = ProfileBuilder()

# Build profile
profile = profile_builder.build(
    df_categorized,
    current_savings=50000,
    current_debt=0,
    goals=[
        {'name': 'Emergency Fund', 'target_amount': 200000, 'months': 12},
        {'name': 'Vacation', 'target_amount': 100000, 'months': 8}
    ]
)

print(profile.summary())
```

```python
# Assess financial health
health = profile_builder.assess_financial_health(profile)

print("\n📊 Financial Health Assessment:")
print(f"Overall Score: {health['overall_score']:.1f}/100")
print(f"Rating: {health['health_rating']}")
print("\nCategory Scores:")
for category, score in health['category_scores'].items():
    print(f"  {category}: {score:.1f}/100")
print("\n📋 Recommendations:")
for rec in health['recommendations']:
    print(f"  • {rec}")
```

## Step 4: Optimize Budget

```python
# Initialize optimizer
optimizer = BudgetOptimizer()

# Optimize budget
optimized_budget = optimizer.optimize(
    income=profile.monthly_income,
    current_spending=profile.expense_categories,
    savings_goal_pct=25
)

print(optimized_budget.summary())
```

```python
# Explain budget recommendations
explainer = Explainer()

explanation = explainer.explain_budget_recommendation(
    current_budget=profile.expense_categories,
    optimized_budget=optimized_budget.category_allocations,
    income=profile.monthly_income
)

print(explanation)
```

## Step 5: What-If Simulations

```python
# Initialize simulator
simulator = WhatIfSimulator(profile.to_dict())

# Simulate car purchase
car_scenario = simulator.simulate_purchase(
    item="Car",
    amount=800000,
    payment_plan="loan",
    duration_months=60,
    interest_rate=8.5
)

print(car_scenario.summary())
```

```python
# Simulate expense reduction
reduction_scenario = simulator.simulate_expense_reduction(
    category="food_dining",
    reduction_percent=30
)

print(reduction_scenario.summary())
```

```python
# Simulate goal achievement
goal_scenario = simulator.simulate_goal(
    goal_name="Vacation Fund",
    target_amount=100000,
    months=8
)

print(goal_scenario.summary())
```

## Step 6: AI Agent Q&A

```python
# Initialize AI agent (requires OpenAI API key)
api_key = os.getenv('OPENAI_API_KEY')

if api_key:
    agent = FinancialAgent(api_key=api_key)
    
    # Initialize with context
    agent.initialize_context(
        profile=profile.to_dict(),
        transactions_summary=patterns,
        budget_plan=optimized_budget.to_dict()
    )
    
    print("✅ AI Agent initialized!")
else:
    print("⚠️ OpenAI API key not found. Set OPENAI_API_KEY in .env file.")
```

```python
# Ask questions to the AI agent
if api_key:
    response = agent.chat("How can I save more money each month?")
    print("\n🤖 Agent Response:")
    print(response)
```

```python
# Get specific recommendations
if api_key:
    savings_rec = agent.get_recommendation("savings")
    print("\n💰 Savings Recommendation:")
    print(savings_rec)
```

## Step 7: Generate Comprehensive Report

```python
# Generate full report
report = explainer.generate_report(
    profile=profile.to_dict(),
    analysis=patterns,
    recommendations=health['recommendations']
)

print(report)

# Save report
with open('../reports/financial_report.txt', 'w') as f:
    f.write(report)
    
print("\n✅ Report saved to reports/financial_report.txt")
```
